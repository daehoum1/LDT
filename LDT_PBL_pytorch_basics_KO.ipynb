### 재현성을 위한 Seed 설정

> **왜 Seed가 필요한가요?**
>
> Seed를 설정하면 실험 결과를 동일하게 재현할 수 있습니다. 이후 딥러닝 실험에서는 다른 사람이 여러분의 코드 실행 결과를 정확히 재현할 수 있도록 하는 것이 매우 중요합니다.

- **난수 생성 결과를 고정하기 위해 Seed 설정**

```python
# Seed 설정
np.random.seed(0)
np.random.rand(2, 2)
```

- **동일한 난수 배열이 생성되는지 다시 확인**

Seed를 설정하지 않으면 여기에서처럼 동일한 숫자들이 반복해서 나오지 않습니다.

```python
# Seed 설정
np.random.seed(0)
np.random.rand(2, 2)
```

- **Seed 없이 NumPy 배열 생성**

앞선 두 번의 실행 결과와 다른 숫자가 나오는 것을 확인할 수 있습니다.

```python
# Seed 미설정
np.random.rand(2, 2)
```

- **Seed 없이 NumPy 배열을 다시 생성**

이번에도 완전히 다른 숫자가 생성되는 것을 확인할 수 있습니다.

```python
# Seed 미설정
np.random.rand(2, 2)
```

- **고정된 Seed를 사용해 PyTorch Tensor 생성**

```python
# PyTorch Seed 설정
torch.manual_seed(0)
torch.rand(2, 2)
```

- **같은 Seed로 PyTorch Tensor를 다시 생성**

```python
# PyTorch Seed 설정
torch.manual_seed(0)
torch.rand(2, 2)
```

- **Seed 없이 PyTorch Tensor 생성**

Seed 없이 NumPy 난수 배열을 생성했을 때와 마찬가지로, 위와 동일한 결과를 얻을 수 없습니다.

```python
# PyTorch Seed 미설정
torch.rand(2, 2)
```

- **Seed 없이 PyTorch Tensor를 다시 생성**

이번에도 숫자가 다시 달라진 것을 확인할 수 있습니다.

```python
# PyTorch Seed 미설정
torch.rand(2, 2)
```

- **GPU의 Seed는 별도로 설정: GPU Tensor를 위한 Seed 고정**

딥러닝 실험에서는 일반적으로 연산 속도를 높이기 위해 GPU를 사용합니다. GPU에서 사용하는 Tensor의 Seed를 고정하는 방법은 위에서 살펴본 CPU의 경우와 다릅니다.

```python
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(0)
```

### CPU와 GPU에서의 Tensor

- **Tensor를 CPU로 이동하고 다시 사용하기**

PyTorch Tensor는 기본적으로 CPU에 생성됩니다. 따라서 CPU에 생성하기 위해 별도의 작업을 할 필요는 없습니다.

```python
# CPU
tensor_cpu = torch.ones(2, 2)
```

Tensor를 GPU로 보내고 싶다면 간단히 `.cuda()`를 사용하면 됩니다.

```python
# CPU에서 GPU로
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
tensor_cpu.to(device)
```

GPU에 있는 Tensor를 다시 CPU로 옮기고 싶다면 다음과 같이 하면 됩니다.

```python
# GPU에서 CPU로
tensor_cpu.cpu()
```

- **추가 예제**

```python
x = torch.randn(2,4)
```

```python
if torch.cuda.is_available():
    device = torch.device("cuda")          # CUDA 디바이스 객체
    y = torch.ones_like(x, device=device)  # GPU에 Tensor를 직접 생성
    x = x.to(device)                       # 또는 문자열 ``.to("cuda")``를 사용
    z = x + y
    print(z)
    print(z.to("cpu", torch.double))       # ``.to``를 사용하면 dtype도 함께 변경 가능
```

## Gradient를 사용하는 Tensor


### Gradient를 사용하는 Tensor 생성
- Gradient를 누적할 수 있도록 설정

- **방법 1: Gradient를 계산하는 Tensor 생성**

일반적인 Tensor 생성 방법과 매우 비슷하며, 추가 인자 하나만 지정하면 됩니다.

```python
a = torch.ones((2, 2), requires_grad=True)
```

- **Tensor가 Gradient 계산을 필요로 하는지 확인**

올바르게 설정했다면 `True`가 반환되어야 합니다.

```python
a.requires_grad
```

- **방법 2: Gradient를 계산하는 Tensor 생성**

먼저 평소처럼 Tensor를 생성한 다음, 추가 코드 한 줄을 통해 Gradient를 누적할 수 있도록 설정하는 방법입니다.

```python
# 일반적인 Tensor 생성 방식
a = torch.ones(2, 2)

# Gradient 계산 활성화
a.requires_grad_()

# Gradient 계산 필요 여부 확인
a.requires_grad
```

- **비교를 위한 Gradient를 사용하지 않는 Tensor**

위의 두 방법 중 어느 것도 적용하지 않으면 Gradient 필요 여부를 확인했을 때 `False`가 반환됩니다.

```python
# Gradient를 사용하지 않는 Tensor
no_gradient = torch.ones(2, 2)
```

```python
no_gradient.requires_grad
```

- **Gradient를 사용하는 Tensor의 덧셈 연산**

```python
# 일반 Tensor와 유사하게 동작
b = torch.ones((2, 2), requires_grad=True)
print(a + b)
print(torch.add(a, b))
```

- **Gradient를 사용하는 Tensor의 곱셈 연산**

앞에서 배운 Tensor 연산은 Gradient를 사용하는 Tensor에도 동일하게 적용됩니다. 나눗셈, 평균, 표준편차 연산도 직접 시도해 보세요.

```python
print(a * b)
print(torch.mul(a, b))
```

### Gradient를 사용하는 Tensor 생성

> **`requires_grad`란 정확히 무엇인가요?**
>
> 해당 Tensor에 대한 Gradient를 계산하고 누적할 수 있도록 설정하는 옵션입니다.

$$y_i = 5(x_i+1)^2$$

- **Gradient 계산이 필요한, 모든 원소가 1인 2x1 Tensor 생성**

```python
x = torch.ones(2, requires_grad=True)
```

```python
x
```

- **생성한 `x` Tensor를 이용한 간단한 식**

$$y_i\bigr\rvert_{x_i=1} = 5(1 + 1)^2 = 5(2)^2 = 5(4) = 20$$

이 간단한 식을 그대로 계산하면 값 20을 얻어야 합니다.

```python
y = 5 * (x + 1) ** 2
```

```python
y
```

- **`y` Tensor를 이용한 간단한 식**

`backward()`는 스칼라(즉, 원소가 1개인 Tensor)에 대해서만 바로 호출할 수 있습니다. 스칼라가 아닌 경우에는 해당 변수에 대한 Gradient를 함께 지정해야 합니다.

따라서 `y`를 스칼라 값으로 줄여 보겠습니다.

$$o = \frac{1}{2}\sum_i y_i$$

위에서 확인한 것처럼 `y`는 모든 원소가 20인 Tensor이므로, 평균을 구하면 20이 됩니다.

```python
o = (1/2) * torch.sum(y)
```

```python
o
```

- **1차 미분값 계산**

**`y` 식 다시 보기**: $y_i = 5(x_i+1)^2$

**`o` 식 다시 보기**: $o = \frac{1}{2}\sum_i y_i$

**`o` 식에 `y`를 대입**: $o = \frac{1}{2} \sum_i 5(x_i+1)^2$

$$\frac{\partial o}{\partial x_i} = \frac{1}{2}[10(x_i+1)]$$

$$\frac{\partial o}{\partial x_i}\bigr\rvert_{x_i=1} = \frac{1}{2}[10(1 + 1)] = \frac{10}{2}(2) = 10$$


따라서 결과로 10이 나올 것으로 예상할 수 있습니다. PyTorch에서는 다음 한 줄로 간단히 계산할 수 있습니다.

1차 미분값 계산:

```python
o.backward()
```

1차 미분값 출력:

```python
x.grad
```

- **`x`가 Gradient를 필요로 하는 경우, 이를 이용해 만든 새로운 객체의 연산도 Gradient 계산 그래프에 포함됩니다**

```python
print(x.requires_grad)
print(y.requires_grad)
print(o.requires_grad)
```

- **Autograd가 Tensor의 연산 기록을 추적하지 않도록 하기**

`.requires_grad=True`로 설정된 Tensor라도 코드 블록을 `with torch.no_grad():`로 감싸면 Autograd가 연산 기록을 추적하지 않도록 할 수 있습니다.

```python
print(x.requires_grad)
print((x ** 2).requires_grad)
```

```python
with torch.no_grad():
    print((x ** 2).requires_grad)
```
