## 예제 1-4. 숫자형 자료를 이용한 계산

````python
a = 10
b = 3

print("더하기:", a + b)
print("빼기:", a - b)
print("곱하기:", a * b)
print("나누기:", a / b)
print("몫:", a // b)
print("나머지:", a % b)
print("제곱:", a ** b)
````

## 예제 1-7. 비교 연산자

비교 연산자는 두 값을 비교하고 결과로 `True` 또는 `False`를 반환합니다.

| 연산자 | 의미 |
|---|---|
| `>` | 크다 |
| `<` | 작다 |
| `>=` | 크거나 같다 |
| `<=` | 작거나 같다 |
| `==` | 같다 |
| `!=` | 같지 않다 |


````python
score = 85

print("점수가 80보다 큰가?", score > 80)
print("점수가 90 이상인가?", score >= 90)
print("점수가 85와 같은가?", score == 85)
print("점수가 100이 아닌가?", score != 100)
````

## 예제 1-8. 논리 연산자

논리 연산자는 여러 조건을 하나로 묶을 때 사용합니다.

| 연산자 | 의미 |
|---|---|
| `and` | 두 조건이 모두 참일 때 참 |
| `or` | 두 조건 중 하나라도 참이면 참 |
| `not` | 참과 거짓을 반대로 바꿈 |


````python
age = 20
has_ticket = True
has_coupon = False
points = 3500

can_enter = age >= 19 and has_ticket
can_discount = has_coupon or points >= 3000

print("입장 가능 여부:", can_enter)
print("할인 가능 여부:", can_discount)
print("쿠폰이 없는가?", not has_coupon)
````

# 2부. 조건문

조건문은 어떤 조건이 참인지 거짓인지에 따라 실행할 코드를 선택하는 문법입니다.

기본 구조는 다음과 같습니다.

```python
if 조건식:
    조건식이 True일 때 실행할 코드
```

파이썬에서는 들여쓰기가 중요합니다.  
`if` 아래에서 실행될 코드는 반드시 일정하게 들여써야 합니다.


## 예제 2-1. 가장 기본적인 if 문

온도가 25도 이상이면 안내 문장을 출력합니다.  
조건이 `False`이면 들여쓰기 된 코드는 실행되지 않습니다.


````python
temperature = 26

if temperature >= 25:
    print("날씨가 따뜻합니다.")
    print("가벼운 옷차림을 준비해도 좋습니다.")

print("이 문장은 조건문 밖에 있으므로 항상 실행됩니다.")
````

## 들여쓰기 오류 예시

아래 코드는 일부러 실행하지 않는 설명용 코드입니다.  
`if` 문 아래 코드를 들여쓰지 않으면 오류가 발생합니다.

```python
temperature = 26

if temperature >= 25:
print("날씨가 따뜻합니다.")  # 들여쓰기가 없어 오류 발생
```

파이썬에서는 중괄호 `{}` 대신 들여쓰기로 코드 블록을 구분합니다.


## 예제 2-2. if / else 문

`else`는 조건이 거짓일 때 실행되는 부분입니다.  
두 가지 경우 중 하나를 선택해야 할 때 사용합니다.


````python
score = 75

if score >= 60:
    print("통과입니다.")
else:
    print("재도전이 필요합니다.")
````

## 예제 2-3. if / elif / else 문

조건이 여러 개일 때는 `elif`를 사용할 수 있습니다.  
위에서 아래로 조건을 검사하고, 가장 먼저 참이 되는 블록만 실행합니다.


````python
score = 84

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "D"

print("점수:", score)
print("등급:", grade)
````

## 예제 2-4. 논리 연산자를 조건문에 사용하기

여러 기준을 함께 판단해야 할 때 `and`, `or`, `not`을 조건문 안에서 사용합니다.


````python
age = 22
has_payment = True
has_coupon = False
points = 2500

if age >= 19 and has_payment:
    print("예약이 가능합니다.")
else:
    print("예약 조건을 확인해야 합니다.")

if has_coupon or points >= 3000:
    print("할인을 받을 수 있습니다.")
else:
    print("할인 조건에 해당하지 않습니다.")
````

## 예제 2-5. 조건문으로 간단한 추천 만들기

조건문은 값을 분류하거나, 상황에 따라 다른 메시지를 출력할 때 자주 사용됩니다.  
아래 예제는 걸음 수에 따라 활동 상태를 나누는 코드입니다.


````python
steps = 7200

if steps >= 10000:
    status = "매우 활발"
elif steps >= 5000:
    status = "보통"
else:
    status = "활동량 부족"

print(f"오늘 걸음 수: {steps}보")
print(f"활동 상태: {status}")
````

# 3부. 반복문

반복문은 같은 작업을 여러 번 실행하는 문법입니다.  
파이썬에서는 대표적으로 `for` 문과 `while` 문을 사용합니다.

- `for`: 정해진 횟수만큼 반복하거나, 목록 안의 값을 하나씩 처리할 때 사용
- `while`: 조건이 참인 동안 계속 반복할 때 사용


## 예제 3-1. for 문과 range()

`range(5)`는 0, 1, 2, 3, 4라는 숫자의 흐름을 만듭니다.  
즉, 5번 반복하지만 숫자는 0부터 시작합니다.


````python
for i in range(5):
    print("현재 i 값:", i)
````

## 예제 3-2. 1부터 10까지 더하기

반복문은 누적 계산에 자주 사용됩니다.  
`total = total + number`는 기존 `total`에 `number`를 더해 다시 `total`에 저장한다는 뜻입니다.


````python
total = 0

for number in range(1, 11):
    total = total + number
    print(f"현재 숫자: {number}, 현재까지 합계: {total}")

print("최종 합계:", total)
````

## 예제 3-3. 리스트를 직접 반복하기

`range()` 대신 리스트를 `for` 문에 넣으면, 리스트 안의 항목이 하나씩 꺼내집니다.


````python
fruits = ["사과", "귤", "포도", "딸기"]

for fruit in fruits:
    print(f"장바구니에 {fruit}가 있습니다.")
````

## 예제 3-4. 반복문과 조건문 함께 사용하기

반복문으로 여러 값을 하나씩 확인하고, 조건문으로 필요한 값만 골라낼 수 있습니다.


````python
scores = [58, 81, 94, 49, 73]

for score in scores:
    if score >= 60:
        print(score, "점: 통과")
    else:
        print(score, "점: 재도전")
````

## 예제 3-5. while 문으로 목표 금액까지 저축하기

`while` 문은 조건이 참인 동안 계속 반복합니다.  
아래 예제에서는 저축액이 50,000원 미만인 동안 10,000원씩 저축합니다.


````python
saving = 0
goal = 50000
week = 0

while saving < goal:
    week = week + 1
    saving = saving + 10000
    print(f"{week}주차 저축액: {saving}원")

print("목표 금액 달성")
````

## 예제 3-6. 무한 반복을 피하는 방법

`while` 문은 조건이 계속 참이면 끝나지 않습니다.  
따라서 반복문 안에서 조건이 언젠가 거짓이 되도록 값을 바꿔야 합니다.

아래 예제에서는 `count`를 1씩 증가시키기 때문에 언젠가 `count < 3`이 거짓이 됩니다.


````python
count = 0

while count < 3:
    print("현재 count:", count)
    count = count + 1

print("반복 종료")
````

# 4부. 함수

함수는 여러 줄의 작업에 이름을 붙여 다시 사용할 수 있게 만든 코드 블록입니다.  
반복되는 코드를 줄이고, 프로그램을 더 읽기 쉽게 만들 때 사용합니다.

기본 구조는 다음과 같습니다.

```python
def 함수이름():
    실행할 코드
```


## 예제 4-1. 가장 간단한 함수

아래 함수는 실행할 때마다 안내 문장 두 줄을 출력합니다.


````python
def show_start_message():
    print("실습을 시작합니다.")
    print("위에서 아래로 천천히 실행하세요.")

show_start_message()
show_start_message()
````

## 예제 4-2. 매개변수와 인자

함수를 만들 때 입력값의 자리 이름을 **매개변수**라고 합니다.  
함수를 사용할 때 실제로 넣는 값을 **인자**라고 합니다.

아래 함수에서 `name`, `age`는 매개변수입니다.  
`"박준영"`, `22`는 함수를 호출할 때 넣는 인자입니다.


````python
def introduce(name, age):
    print(f"이름은 {name}입니다.")
    print(f"나이는 {age}세입니다.")

introduce("박준영", 22)
introduce("김서연", 20)
````

## 예제 4-3. return으로 결과 돌려주기

`return`은 함수 안에서 만든 결과를 함수 밖으로 돌려줍니다.  
함수의 결과를 다른 계산에 계속 사용하려면 `return`이 필요합니다.


````python
def add(a, b):
    result = a + b
    return result

answer = add(12, 18)
print("함수에서 반환된 값:", answer)
print("반환값에 5를 더한 값:", answer + 5)
````

## 예제 4-4. print와 return의 차이

`print()`는 화면에 보여주는 기능입니다.  
`return`은 값을 함수 밖으로 전달하는 기능입니다.

둘은 비슷해 보이지만 역할이 다릅니다.


````python
def add_and_print(a, b):
    print(a + b)


def add_and_return(a, b):
    return a + b

printed_result = add_and_print(2, 3)
returned_result = add_and_return(2, 3)

print("print만 사용한 함수의 결과:", printed_result)
print("return을 사용한 함수의 결과:", returned_result)
````

`print`만 사용한 함수는 화면에는 값을 보여주지만, 함수 밖으로 실제 계산 결과를 넘기지는 않습니다.  
그래서 `printed_result`에는 `None`이 저장됩니다.


## 예제 4-5. 기본값 매개변수

기본값 매개변수는 함수 호출 시 값을 생략했을 때 자동으로 사용되는 값입니다.


````python
def greet(name, message="안녕하세요"):
    print(f"{name}님, {message}.")

# message를 생략하면 기본값이 사용됩니다.
greet("박준영")

# message를 직접 넣으면 직접 넣은 값이 사용됩니다.
greet("김서연", "좋은 아침입니다")
````

## 예제 4-6. 함수로 평균과 등급 계산하기

함수는 계산 규칙을 하나로 묶을 때 유용합니다.  
아래 예제에서는 평균을 계산하는 함수와 등급을 판단하는 함수를 따로 만듭니다.


````python
def calculate_average(scores):
    total = sum(scores)
    count = len(scores)
    return total / count


def make_grade(average):
    if average >= 90:
        return "A"
    elif average >= 80:
        return "B"
    elif average >= 70:
        return "C"
    else:
        return "D"

scores = [82, 90, 86]
average = calculate_average(scores)
grade = make_grade(average)

print("점수 목록:", scores)
print("평균:", average)
print("등급:", grade)
````

## 연습 4. 함수 입력값 바꿔 보기

아래 코드에서 `prices` 값을 바꿔 실행해 보세요.  
합계와 평균이 함께 바뀝니다.


````python
def summarize_prices(prices):
    total = sum(prices)
    average = total / len(prices)
    return total, average

prices = [1500, 2400, 3200, 1900]
total_price, average_price = summarize_prices(prices)

print("가격 목록:", prices)
print("합계:", total_price)
print("평균:", average_price)
````

# 8부. 객체 지향 프로그래밍 맛보기

파이썬에서 리스트, 문자열, 딕셔너리 같은 자료형은 모두 객체로 다룰 수 있습니다.  
객체는 크게 두 가지를 가질 수 있습니다.

- **속성**: 객체가 가지고 있는 정보
- **메서드**: 객체가 수행할 수 있는 기능

예를 들어 리스트의 `append()`는 리스트 객체가 제공하는 메서드입니다.


## 예제 8-1. 이미 사용하고 있던 메서드 확인하기

문자열, 리스트, 딕셔너리에는 각각 자주 사용하는 메서드가 있습니다.


````python
text = "python"
print("대문자로 변환:", text.upper())
print("첫 글자만 대문자로:", text.capitalize())

numbers = [4, 2, 1]
numbers.append(3)
print("append 후 리스트:", numbers)

student = {"name": "박준영", "score": 91}
print("딕셔너리 키:", student.keys())
````

## 예제 8-2. 클래스와 객체

클래스는 어떤 정보와 기능을 가질지 정해 둔 틀입니다.  
객체는 그 틀을 바탕으로 실제로 만들어진 하나의 데이터입니다.

아래 예제에서는 `Student`라는 클래스를 만들고, 학생 객체를 생성합니다.


````python
class Student:
    def __init__(self, name, scores):
        # self.name과 self.scores는 객체가 가지는 속성입니다.
        self.name = name
        self.scores = scores

    def average(self):
        # average는 객체가 수행할 수 있는 메서드입니다.
        return sum(self.scores) / len(self.scores)

    def show_summary(self):
        print(f"{self.name} 학생의 평균 점수는 {self.average():.1f}점입니다.")

student_a = Student("박준영", [82, 91, 86])
student_b = Student("김서연", [94, 89, 93])

student_a.show_summary()
student_b.show_summary()
````

## 예제 8-3. 객체마다 다른 값 가질 수 있음

같은 `Student` 클래스로 만든 객체라도 내부 속성 값은 서로 다를 수 있습니다.  
이 점이 클래스와 객체를 구분할 때 중요합니다.


````python
print("student_a의 이름:", student_a.name)
print("student_a의 점수:", student_a.scores)
print("student_a의 평균:", student_a.average())

print("student_b의 이름:", student_b.name)
print("student_b의 점수:", student_b.scores)
print("student_b의 평균:", student_b.average())
````

## NumPy 전에 객체 지향을 살짝 보는 이유

다음 단계에서 사용할 NumPy 배열도 단순한 숫자 묶음이 아니라 하나의 객체처럼 다루어집니다.  
배열은 다음과 같은 정보를 가지고, 여러 기능도 제공합니다.

- 모양: 몇 행, 몇 열인지
- 자료형: 정수인지, 실수인지
- 크기: 원소가 몇 개인지
- 기능: 합계, 평균, 형태 변경 등

지금은 클래스를 깊게 외우기보다, **데이터와 기능이 하나의 대상에 묶일 수 있다**는 감각을 잡는 것이 중요합니다.
